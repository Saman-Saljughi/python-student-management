```python
import json
with open("students.json", "r")as file:
    students= json.load(file)
def average_score(scores):
    if len(scores)==0:
        return None
    total=0
    count=0
    for score in scores:
        total+=score
        count+=1
    return total/count

def student_average(name,students):
    scores= students[name]["scores"]
    average= average_score(scores)
    return average

def find_best_student(students):
    best_name= None
    best_average= None
    for name in students:
        average= student_average(name, students)
        if average is None:
            continue
        if best_name is None:
            best_name= name
            best_average= average
        elif average > best_average :
                best_name= name
                best_average= average
    return best_name , best_average
    
            
def find_weakest_student(students):
    lowest_average= None
    weakest_name= None
    for name in students:
        average= student_average(name, students)
        if average is None:
            continue
        if weakest_name is None:
            weakest_name= name
            lowest_average= average
        if average < lowest_average:
            lowest_average= average
            weakest_name= name
    return weakest_name ,lowest_average 
    
def class_average(students):
    averages= 0
    count= 0
    for name in students:
        average= student_average(name, students)
        if average is None:
            continue
        averages+= average
        count+= 1
    if count==0:
        return None
    return averages / count
    
def averages_above_average(students):
    averages_up_average= []
    class_averages=class_average(students)
    if class_averages is None:
        return []
    for name in students:
        average= student_average(name, students)
        if average is None:
            continue
        elif average >= class_averages:
            averages_up_average.append((name,average))
    return averages_up_average

def show_student(name):
    if name in students:
        average= student_average(name, students)
        print(" دانش آموز یافت شد✅")
        print(" نام:", name)
        print("سن:", students[name]["age"])
        print(" نمرات:", students[name]["scores"])
        print(" معدل:", average)
    else:
        print("دانش آموز یافت نشد😴❌")

def show_all_students(students):
    for name in students:
        show_student(name)

username=input("لطفا نام کاربری خود را وارد کنید")
password=input("لطفا پسورد خود را وارد کنید")
if username == "Saman" and password == "Nana":
    print("ورود با موفق🤝🏻✅")
    while True:
        print("1.نمایش اطلاعات دانش آموز📝")
        print("2.ویرایش سن و نمره✏️")
        print("3.نمایش گزارش ها📋")
        print("4.اضافه کردن دانش آموز جدید")
        print("5.خروج➡️🚪")
        choice=input("لطفا عدد گزینه مورد نظر را وارد کنید🖌️")
        if choice=="5":
            break
        
        elif choice=="1":
            print("به منوی اطلاعات دانش آموز وارد شدید✅")
            print("________________________________________")
            print(">1<","نمایش اطلاعات دانش آموز")
            print(">2<","نمایش اطلاعات همه دانش آموزان")
            show_choice=input("لطفا انتخاب کنید✏️")
            if show_choice=="1":
                name=input("لطفا نام دانش آموز را وارد کنید✏️")
                show_student(name)
            elif show_choice=="2":
                show_all_students(students)
            else:
                print("انتخاب نامعتبر❌")
                
        elif choice=="2":
            print("به منوی ویرایش سن و نمره وارد شدید✅")
            name=input("لطفا نام دانش آموز را وارد کنید")
            if name in students:
                print("»1«","تغییر سن")
                print("»2«","تغییر نمره")
                edit_choice=input("لطفا انتخاب کنید")
                if edit_choice=="1":
                    try:
                        new_age=int(input("لطفا سن جدید را وارد کنید✏️"))
                    except ValueError:
                        print("لطفا فقط عدد وارد کنید❌")
                    else:
                        if new_age>=0:
                            students[name]["age"]=new_age
                            with open("students.json","w")as file:
                                json.dump(students,file,indent=4)
                            print("✅سن با موفقیت ثبت شد")
                        else:
                            print("سن نامعتبر❌")
                            print("سن نباید منفی باشد❌")
                elif edit_choice=="2":
                    try:
                        new_score=int(input("لطفا نمره را وارد کنید"))
                    except ValueError:
                        print("لطفا فقط عدد وارد کنید❌")
                    else:
                        if 0<= new_score <=20:
                            students[name]["scores"].append(new_score)
                            with open("students.json","w")as file:
                                json.dump(students,file,indent=4)
                            print("نمره جدید با موفقیت وارد شد✅")
                        else:
                            print("نمره نامعتبر❌")
                            print("عدد باید بین 0 تا 20 باشد")
                else:
                    print("انتخاب نا معتبر❌😒")
            else:
                print("دانش آموز یافت نشد❌😴")
        
        elif choice=="3":
            print("به منوی نمایش گزارش ها وارد شدید✅")
    
            best_student, best_average = find_best_student(students)
            if best_student is None:
                print("❌هنوز نمره ای برای محاسبه بهترین معدل وجود ندارد")
            else:
                print(best_student, "بهترین دانش آموز🥇🏆")
                print(best_average, "بهترین میانگین🎖️")
            print("___________________________________________")
            weakest_student , weakest_average= find_weakest_student(students)
            if weakest_student is None:
                print("هنوز نمره ای برای محاسبه ضعیف ترین معدل وجود ندارد❌")
            else:
                print(weakest_student, "بدترین دانش آموز🫵🏻🤣😴")
                print(weakest_average, "کمترین معدل⚠️😕")
            print("___________________________________________")
            class_avg= class_average(students)
            avg_up_avg=averages_above_average(students)
            if class_avg is None:
                print("❌هنوز نمره ای برای محاسبه معدل کلاس وجود ندارد")
                print(avg_up_avg,"هنوز نمره ای برای پیدا کردن معدل های بالاتر از میانگین کلاس وجود ندارد❌")
            else:
                print(class_avg,"میانگین کلاس🤝🏻🎗️")
                print("________________________________________")
            
            if not avg_up_avg:
                print("دانش آموزی برای نمایش وجود ندارد")
            else:
                pprint(avg_up_avg,"معدل های بلاتر از میانگین کلاس🏆")
        
        elif choice=="4":
            print("وارد منوی اضافه کردن دانش آموز جدید شدید✅")
            name=input("لطفا نام دانش آموز جدید را وارد کنید✏️")
            if name in students:
                print("این نام از قبل در سیستم ثبت شده است؛ لطفاً نام دیگری وارد کنید❌")
            if name not in students:
                print("نام قابل ثبت است✅")
                try:
                    new_age=int(input("سن را وارد کنید✏️"))
                except ValueError:
                    print("لطفا فقط عدد وارد کنید❌")
                else:
                    if new_age>0:
                        new_student={
                                "age":new_age,
                                "scores":[]
                            }
                        new_name=name
                        students[name]=new_student
                        students[new_name]["age"]=new_age
                        with open("students.json","w")as file:
                            json.dump(students,file,indent=4)
                        print("دانش آموز جدید با موفقیت ثبت شد✅")
            
        else:
            print("گزینه نامعتبر است❌")
else:
    print("نام کاربری یا پسورد نامعتبر❌😴")
```
